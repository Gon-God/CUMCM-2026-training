```python
import scipy.cluster.hierarchy as sch
import numpy as np
import pylab as plt

import matplotlib.pyplot as plt
plt.rcParams['text.usetex'] = False   # 强制使用 Matplotlib 内置渲染器

#plt.rc('text',usetex=True)
plt.rc('font')

a = np.array([[2,3,3.5,7,9]]).T
c = sch.linkage(a, method='single')
s = ['$\\omega_'+str(i+1)+'$' for i in range(5)]
dendrogram = sch.dendrogram(c,labels=s)
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
    [0.294, 1.093, 5.63, 113.6, 4510, 1036, 12.2],
    [0.315, 0.971, 0.39, 95.1, 2773, 683, 0.85],
    [0.123, 0.316, 5.28, 148.5, 6934, 611, 6.49],
    [0.179, 0.527, 0.39, 111, 4458, 632, 0.92],
    [0.081, 0.212, 72.04, 217.8, 12240, 791, 80.3],
    [0.082, 0.211, 43.78, 179.6, 8973, 636, 48.1],
    [0.075, 0.181, 65.15, 194.7, 1068, 634, 80.1],
    [0.293, 0.666, 5.35, 94.9, 3679, 771, 7.8],
    [0.167, 0.414, 2.9, 94.8, 4231, 574, 1.17]
])



b = (data-data.min(axis=0))/(data.max(axis=0)-data.min(axis=0) ) #get the normalized data
c = sch.linkage(b, method='complete')
s = ['$\\omega_'+str(i+1)+'$' for i in range(len(data))]
dendrogram = sch.dendrogram(c,labels=s)
plt.show()
```

```python
from sklearn.cluster import KMeans
import numpy as np

a = np.array([[1,3],[2,4],[3,5],[8,10],[9,11]])
md = KMeans(n_clusters=2).fit(a)
labels = md.labels_
centers = md.cluster_centers_
print(labels)
print(centers)
```

```python
# 如果尚未安装 scikit-learn，先安装（通常 Anaconda 已自带）
# conda install scikit-learn  或  pip install scikit-learn

from sklearn.cluster import KMeans
import numpy as np

a = np.array([[1,3],[2,4],[3,5],[8,10],[9,11]])
kmeans = KMeans(n_clusters=2, random_state=0).fit(a)
print(kmeans.labels_)          # 聚类标签
print(kmeans.cluster_centers_) # 聚类中心
```

```python
from sklearn.cluster import KMeans
import pylab as plt
import numpy as np

a = np.array([[1,3],[2,4],[3,5],[8,10],[9,11]])
md = []
g = np.arange(1,len(a)+1)
for i in range(1,len(a)+1):
	md.append(KMeans(n_clusters=i).fit(a).inertia_)

plt.plot(g,md,'*-')
plt.show()
```
